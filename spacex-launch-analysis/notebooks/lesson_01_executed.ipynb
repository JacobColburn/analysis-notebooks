## Step 1: Import pandas

`pandas` is the Swiss Army knife of data analysis in Python. It gives us a structure called a **DataFrame** — think of it as a spreadsheet that lives in Python.

We import it with the alias `pd` — you'll see this convention everywhere.

```python
import pandas as pd

# This just silences a common harmless warning
pd.set_option('display.max_columns', None)

print('pandas version:', pd.__version__)
print('✅ pandas loaded successfully')
```

## Step 2: Load the dataset

`pd.read_csv()` reads a CSV file and turns it into a DataFrame.

Our dataset has **135 SpaceX launches** from 2006–2024, with columns for:
- `flight_number` — sequential launch number
- `name` — mission name
- `date` — launch date
- `rocket` — rocket type (Falcon 1, Falcon 9, Falcon Heavy, Starship)
- `launchpad` — where it launched from
- `success` — did the mission succeed? (True/False)
- `payload_kg` — payload mass in kilograms
- `orbit` — target orbit
- `landing_success` — did the booster land? (True/False/None)
- `reused` — was this a reused booster?
- `customer_type` — Government or Commercial

```python
# Load the CSV — adjust the path if needed
df = pd.read_csv('../data/spacex_launches.csv')

print(f'Dataset loaded: {df.shape[0]} rows × {df.shape[1]} columns')
```

## Step 3: First look at the data

`.head()` shows you the first 5 rows. This is always your first move — just *look* at what you're working with.

```python
df.head()
```

```python
# .tail() shows the LAST 5 rows — useful to see the most recent data
df.tail()
```

## Step 4: Understand the columns

`.info()` is one of the most useful commands in pandas. It tells you:
- Column names
- Data types (int, float, bool, object = string)
- How many non-null (non-missing) values each column has

```python
df.info()
```

**🧠 Notice anything?** The `landing_success` column has fewer non-null values than the others. That means some rows have missing data there. This makes sense — early Falcon 1 launches didn't even attempt a landing.

We'll deal with missing data in a later lesson. For now, just be aware it's there.

## Step 5: Check for missing values

`.isnull().sum()` counts how many missing values exist in each column.

```python
print('Missing values per column:')
print(df.isnull().sum())
```

## Step 6: Summary statistics

`.describe()` gives you count, mean, min, max, and percentiles for every numeric column. It's a quick pulse check.

```python
df.describe()
```

**🧠 Key observations from `.describe()`:**
- `payload_kg` ranges from 0 to ~18,500 kg — huge range, because early missions were tiny and Starlink batches are massive
- `flight_number` goes from 1 to 135 — that's our full timeline

---

## Step 7: Ask your first questions

Now we start *actually* exploring. Let's answer some basic questions.

```python
# Q1: What rockets has SpaceX launched?
print('Rocket types and launch counts:')
print(df['rocket'].value_counts())
```

```python
# Q2: What is SpaceX's overall mission success rate?
total = len(df)
successes = df['success'].sum()  # True counts as 1
rate = successes / total * 100

print(f'Total launches:     {total}')
print(f'Successful:         {successes}')
print(f'Failed:             {total - successes}')
print(f'Success rate:       {rate:.1f}%')
```

```python
# Q3: What orbit types have they targeted?
print('Target orbits:')
print(df['orbit'].value_counts())
```

```python
# Q4: Government vs Commercial — who's the bigger customer?
print('Customer breakdown:')
print(df['customer_type'].value_counts())
print()
print('As percentages:')
print(df['customer_type'].value_counts(normalize=True).mul(100).round(1).astype(str) + '%')
```

```python
# Q5: What year had the most launches?

# First, convert the 'date' column to a proper datetime type
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year  # Extract the year into its own column

print('Launches per year:')
print(df['year'].value_counts().sort_index())
```

```python
# Q6: How does reuse look over time?
reuse_by_year = df.groupby('year')['reused'].mean().mul(100).round(1)
print('% of launches using reused boosters by year:')
print(reuse_by_year)
```

## 🎯 Lesson 1 Summary

You just did real exploratory data analysis. Here's what we covered:

| Command | What it does |
|---|---|
| `pd.read_csv()` | Load a CSV into a DataFrame |
| `df.shape` | Rows × columns |
| `df.head()` / `df.tail()` | Peek at the data |
| `df.info()` | Column types + missing value count |
| `df.isnull().sum()` | Count missing values |
| `df.describe()` | Summary statistics |
| `df['col'].value_counts()` | Count unique values in a column |
| `df.groupby()` | Group rows and aggregate |
| `pd.to_datetime()` | Convert strings to proper dates |

**Key findings so far:**
- SpaceX's overall success rate is very high (>90%)
- The Falcon 9 Block 5 is their workhorse
- Booster reuse went from 0% to near-100% over time — a massive operational shift
- LEO and GTO are the dominant target orbits
