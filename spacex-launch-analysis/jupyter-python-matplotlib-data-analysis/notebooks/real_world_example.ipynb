```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/spacex_launches.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
# how many failures
df[df['success'] == False]
```

```python
# are all the failures early on?
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df[df['success'] == False][['name', 'year', 'rocket']]
```

```python
# yeah mostly early - check success rate by year
df.groupby('year')['success'].mean().mul(100).round(1)
```

```python
# plot it
df.groupby('year')['success'].mean().mul(100).plot(marker='o')
plt.title('success rate by year')
plt.ylabel('%')
plt.show()
```

```python
# curious about the 2015 dip - what happened
df[(df['year'] == 2015) & (df['success'] == False)][['name', 'rocket']]
```

```python
# CRS-7 exploded on ascent - check if reused boosters have anything to do with failures
df.groupby('reused')['success'].mean().mul(100).round(1)
```

```python
# interesting - reused actually slightly better. check sample sizes
df.groupby('reused')['success'].agg(['mean', 'count'])
```
